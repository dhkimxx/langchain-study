# trafilatura POC: Web Page 코드블럭 추출 테스트

- 목표: trafilatura로 본문을 추출했을 때 코드블럭(markdown ``` 블록)이 얼마나 잘 보존되는지 확인

아래 셀을 순서대로 실행하면서 결과를 확인합니다.


````python
import trafilatura

# URL = "https://toss.tech/article/payments-legacy-3"
URL = "https://velog.io/@mgang0_0/useRef%EB%9E%80-%EB%AC%B4%EC%97%87%EC%9D%BC%EA%B9%8C"

downloaded = trafilatura.fetch_url(URL)
print("fetched:", bool(downloaded), "length:", len(downloaded) if downloaded else 0)

text_markdown = trafilatura.extract(
    downloaded,
    url=URL,
    favor_precision=True,
    output_format="markdown",
)

print("\n=== 추출된 전체 본문 길이 ===")
print(len(text_markdown) if text_markdown else 0)
print("\n=== 앞부분 800자 프리뷰 ===\n")
print(text_markdown[:800] if text_markdown else "(empty)")
````

````python
import re

if not text_markdown:
    raise ValueError("먼저 위 셀을 실행해 text_markdown을 생성해 주세요.")

code_blocks = re.findall(r"```.*?```", text_markdown, flags=re.DOTALL)

print(f'코드블럭 개수: {len(code_blocks)}')

for i, block in enumerate(code_blocks, 1):
    print(f'\n===== Code block {i} =====\n')
    print(block)
    print('\n==========================')
````
